````python
arg1 = "```json\n[\n  {\n    \"chapter\": 1,\n    \"subchapter\": 1\n  },\n  {\n    \"chapter\": 1,\n    \"subchapter\": 2\n  },\n  {\n    \"chapter\": 2,\n    \"subchapter\": 1\n  },\n  {\n    \"chapter\": 2,\n    \"subchapter\": 2\n  },\n  {\n    \"chapter\": 3,\n    \"subchapter\": 1\n  },\n  {\n    \"chapter\": 3,\n    \"subchapter\": 2\n  },\n  {\n    \"chapter\": 4,\n    \"subchapter\": 1\n  },\n  {\n    \"chapter\": 4,\n    \"subchapter\": 2\n  },\n  {\n    \"chapter\": 5,\n    \"subchapter\": 1\n  },\n  {\n    \"chapter\": 5,\n    \"subchapter\": 2\n  }\n]\n```"
````

````python
def main(arg1: str) -> dict:
    import json
    # print(arg1)
    arg1 = arg1.replace('```json', '')
    arg1 = arg1.replace('```', '')
    print(arg1)
    data = json.loads(arg1)
    
    # Create an array of objects
    print(data)
    result = [{'chapter': item["chapter"], 'subchapter': item["subchapter"]} for item in data]
    
    return {
        'result': result
    }


main(arg1)
````
